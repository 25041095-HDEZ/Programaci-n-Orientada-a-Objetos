```python
# En este ejemplo, todo esta bien. Se espera un numero, el usuario introduce un
# numero y se realiza el proceso como debe ser
x = float(input("introduce un numero"))
y = 5
print("La multiplicacion de", y, "por", x, "es=", x*y , "\n")
```

```python
while(True):
  try:
    x = float(input("introduce un numero"))
    y = 5
    print("La multiplicacion de", y, "por", x, "es=", x*y , "\n")
    break
  except:
    print("Ha ocurrido un error")
```

```python
# En este ejemplo
while(True):
  try:
    x = float(input("introduce un numero "))
    y = 5
    print("La multiplicacion de", y, "por", x, "es=", x*y , "\n")
  except:
    print("Ha ocurrido un error")
  else:
    print("Todo funciona bien")
    break
  finally:
    print("Codigo final")
```

```python
while(True):
  try:
    x = float(input("introduce un numero "))
    y = 100
    print("La division de", y, "entre", x, "es=", y/x)
  except ValueError:
    print("No es numerico el valor proporcionado")
  except ZeroDivisionError:
    print("No puedes dividir entre 0")
  except Exception as e:
    print("Error", type(e) .__name__)
  else:
    print("Todo funciona bien")
    break
  finally:
    print("Codigo final \n")

  print("Fuera del try")
```
