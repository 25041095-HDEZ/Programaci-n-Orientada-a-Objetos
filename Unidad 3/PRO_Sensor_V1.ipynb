```python
class GPIO:
    def __init__(self, output, input, sensor):
     self.output = output
     self.input = input
     self.sensor = sensor

    def info(self):
       print(f"Salidas de la RamsBerry: {self.output} | Entradas de la RamBerry: {self.input} | Sensor: {self.sensor}")

class Sensor(GPIO):
    def __init__(self, pins, output, input, sensor):
       super().__init__(output, input, sensor)
       self.pins = pins

    def info(self):
        print(f"Pines del Sensor: {self.pins} | Sensor: {self.sensor}")
```
